## Poređenje R² score-a najboljih modela

```python
model_names = [
    'Linearna regresija',
    'Linearni SVM',
    'Kernelizovani SVM (RBF)',
    'KNN',
    'Random Forest',
    'Neuronska mreža'
]
r2_scores = [
    r2_ElasticNet,
    r2_LinearSVM,
    r2_SVM_kernel,
    r2_KNN,
    r2_RandomForest,
    r2_NeuronskaMreza
]

sorted_pairs = sorted(zip(r2_scores, model_names))
sorted_scores = [s for s, _ in sorted_pairs]
sorted_names = [n for _, n in sorted_pairs]

plt.figure(figsize=(10, 6))
bars = plt.barh(sorted_names, sorted_scores, color='steelblue')
plt.xlabel('R² score')
plt.title('Poređenje R² scora-a najboljih modela po tipu')
plt.xlim(0, 1)

bars[-1].set_color('seagreen')

for bar, score in zip(bars, sorted_scores):
    plt.text(bar.get_width() + 0.01, bar.get_y() + bar.get_height()/2,
             f'{score:.3f}', va='center')

plt.tight_layout()
plt.show()
```

Poređenjem R² skora najboljih varijanti svakog modela stiče se jasan uvid u to koliko različiti pristupi uspevaju da objasne varijansu u podacima. Ubedljivo najbolje rezultate ostvaruje Random Forest sa R² vrednošću od 0.799, čime objašnjava blizu 80% varijanse ciljne promenljive. Slede ga neuronska mreža sa 0.697 i kernelizovani SVM sa RBF kernelom sa 0.652, dok se KNN sa 0.619 nalazi neposredno iza njih. Ova četiri modela dele zajedničku osobinu da mogu da modeluju nelinearne odnose među atributima, što se pokazalo ključnim za uspešno predviđanje.

Na suprotnom kraju nalaze se linearni modeli, koji ostvaruju znatno slabije performanse. Linearna regresija ( ElasticNet ) postiže R² od 0.268, a linearni SVM svega 0.194. Ovako niske vrednosti ukazuju na to da linearna pretpostavka nije adekvatna za posmatrani problem, jer takvi modeli nisu u stanju da uhvate složene, nelinearne zavisnosti prisutne u podacima.
